```python
import yfinance as yf
import pandas as pd
import os

def get_financial_data(tickers, filename="market_data.csv"):
    """
    Veriyi bir kez indirir ve CSV olarak saklar. 
    Tekrar çalıştırıldığında yerelden okur.
    """
    if os.path.exists(filename):
        print(f"--- Veri yerel dosyadan ({filename}) yükleniyor... ---")
        # Tarih sütununu indeks olarak ayarla
        data = pd.read_csv(filename, index_col=0, parse_dates=True)
    else:
        print(f"--- Veri yfinance üzerinden indiriliyor... ---")
        # Veriyi indir
        raw_data = yf.download(tickers, period="5y", interval="1d")
        
        # Sadece 'Close' fiyatlarını al ve CSV'ye kaydet
        data = raw_data['Close']
        data.to_csv(filename)
        print(f"--- Veri başarıyla '{filename}' adıyla kaydedildi. ---")
    
    return data

# Kullanım örneği:
stocks = ['AAPL', 'TSLA', 'JPM', 'XOM', 'JNJ', 'WMT']
indices = {'S&P500': 'SPY', 'NASDAQ': 'QQQ', 'DOW': 'DIA'}
all_tickers = stocks + list(indices.values())

# Fonksiyonu çağır
data_close = get_financial_data(all_tickers)
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from statsmodels.stats.diagnostic import acorr_ljungbox
from arch import arch_model   # <-- YENİ (GARCH için)

warnings.filterwarnings('ignore', category=UserWarning)

stocks = ['AAPL', 'TSLA', 'JPM', 'XOM', 'JNJ', 'WMT']
indices = {'S&P500': 'SPY', 'NASDAQ': 'QQQ', 'DOW': 'DIA'}
all_tickers = stocks + list(indices.values())

# -------------------------------------------------------------
# GARCH filtresi
# -------------------------------------------------------------
def apply_garch_filter(series):
    scaled = series * 100
    try:
        # AR(1) - GARCH(1,1) modeli (Finansal getirilerdeki bağımlılığı IID yapmak için standarttır)
        model = arch_model(scaled, mean='AR', lags=1, vol='GARCH', p=1, q=1, dist='StudentST')
        res = model.fit(disp='off', show_warning=False)
        std_resid = res.std_resid
        return std_resid.dropna().values
    except Exception:
        s = series.values
        return (s - np.mean(s)) / np.std(s)

# -------------------------------------------------------------
# Ana analiz fonksiyonu (değiştirilen kısım)
# -------------------------------------------------------------
def financial_analysis_pipeline(csv_path):
    data = pd.read_csv(csv_path, index_col=0, parse_dates=True)
    returns = np.log(data / data.shift(1)).dropna()

    # 1) Tüm serilere GARCH filtresi uygula
    filtered = {}
    print("Applying ARMA-GARCH filtering...")
    for ticker in all_tickers:
        if ticker in returns.columns:
            filtered[ticker] = apply_garch_filter(returns[ticker])

    results_list = []

    for stock in stocks:
        # Ljung-Box testi artık GARCH kalıntıları üzerinde (Bağımsızlık kontrolü)
        if stock in filtered and len(filtered[stock]) > 10:
            lb_p_value = acorr_ljungbox(filtered[stock], lags=[10], return_df=True).iloc[0,1]
        else:
            lb_p_value = np.nan

        for idx_name, idx_ticker in indices.items():
            if stock not in filtered or idx_ticker not in filtered:
                continue

            s_std = filtered[stock]
            i_std = filtered[idx_ticker]
            
            # s_std ve i_std kesimi aynı kalıyor:
            min_len = min(len(s_std), len(i_std))
            s_std = s_std[:min_len]
            i_std = i_std[:min_len]

            # DÜZELTİLDİ: Orijinal getirilerin sondan min_len kadarını al (GARCH ilk günleri yuttuğu için)
            s_orig = returns[stock].values[-min_len:]
            i_orig = returns[idx_ticker].values[-min_len:]

            corr = np.corrcoef(s_orig, i_orig)[0, 1]
            ann_vol = np.std(s_orig) * np.sqrt(252)

            # ------------- GOF Testleri (GARCH kalıntıları üzerinde) -------------
            ks_p = stats.ks_2samp(s_std, i_std).pvalue
            try:
                ad_res = stats.anderson_ksamp([s_std, i_std],
                                              method=stats.PermutationMethod(n_resamples=1000))
                ad_p = ad_res.pvalue
            except:
                ad_p = stats.anderson_ksamp([s_std, i_std]).pvalue
            cvm_p = stats.cramervonmises_2samp(s_std, i_std).pvalue
            try:
                es_p = stats.epps_singleton_2samp(s_std, i_std)[1]
            except:
                es_p = np.nan

            sw_p = stats.shapiro(s_std).pvalue

            results_list.append({
                'Stock': stock,
                'Index': idx_name,
                'Corr_R': corr,
                'Ann_Vol': ann_vol,
                'LB_p': lb_p_value,
                'KS_p': ks_p,
                'AD_p': ad_p,
                'CvM_p': cvm_p,
                'ES_p': es_p,
                'Normality_p': sw_p
            })

    return pd.DataFrame(results_list), returns

# -------------------------------------------------------------
# Aşağıdaki tüm fonksiyonlar ve çalıştırma kısmı HİÇ DEĞİŞMEDİ
# -------------------------------------------------------------
def format_sci_values(val, is_p_value=True):
    if pd.isna(val):
        return "N/A"
    if is_p_value:
        star = ""
        if val < 0.01: star = "***"
        elif val < 0.05: star = "**"
        elif val < 0.10: star = "*"
        if val < 0.001:
            return f"<.001{star}"
        formatted = f"{val:.3f}"
        if 0 <= val < 1:
            formatted = formatted.replace("0.", ".")
        return f"{formatted}{star}"
    else:
        return f"{val:.3f}"

def plot_comprehensive_visuals(df_results, returns_data):
    sns.set_theme(style="whitegrid")
    # Figure 1: Cumulative returns
    plt.figure(figsize=(12, 6), dpi=300)
    cum_returns = returns_data.cumsum()
    for stock in stocks:
        plt.plot(cum_returns.index, cum_returns[stock], label=stock, alpha=0.5)
    plt.plot(cum_returns.index, cum_returns['SPY'], color='black', lw=3, label='S&P 500')
    plt.plot(cum_returns.index, cum_returns['QQQ'], color='blue', lw=2, ls='--', label='NASDAQ')
    plt.plot(cum_returns.index, cum_returns['DIA'], color='red', lw=2, ls=':', label='DOW')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig('figure1_cumulative_returns.png', dpi=300, bbox_inches='tight')
    plt.show()
    # Figure 2: Heatmap
    test_cols = ['KS_p', 'AD_p', 'CvM_p', 'ES_p']
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    axes = axes.flatten()
    for i, col in enumerate(test_cols):
        pivot_df = df_results.pivot(index="Stock", columns="Index", values=col)
        sns.heatmap(pivot_df, annot=True, cmap="RdYlGn", center=0.05, fmt=".3f", ax=axes[i])
        axes[i].set_title(f'Test: {col.replace("_p", "")} Similarity')
    plt.tight_layout()
    plt.savefig('fig2_stg1_heatmap.png', dpi=300, bbox_inches='tight')
    plt.show()
    # Figure 3: Correlation matrix
    plt.figure(figsize=(10, 8))
    corr_matrix = returns_data[all_tickers].corr()
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
    sns.heatmap(corr_matrix, mask=mask, annot=True, cmap='coolwarm', fmt=".2f")
    plt.tight_layout()
    plt.savefig('fig3_corr.png', dpi=300, bbox_inches='tight')
    plt.show()
    # Correlation table
    print("\n" + "="*50)
    print("COMPLETE PEARSON CORRELATION MATRIX (Stocks & Benchmarks)")
    print("="*50)
    full_corr = returns_data[all_tickers].corr()
    print(full_corr.to_markdown(floatfmt=".3f"))
    print("\n" + "="*50)
    # Figure 4: Rolling volatility
    plt.figure(figsize=(12, 6))
    rolling_vol = returns_data.rolling(window=252).std() * np.sqrt(252)
    for stock in stocks:
        plt.plot(rolling_vol.index, rolling_vol[stock], label=f'{stock}', alpha=0.3, lw=2)
    plt.plot(rolling_vol.index, rolling_vol['SPY'], color='black', lw=3, label='S&P 500')
    plt.plot(rolling_vol.index, rolling_vol['QQQ'], color='blue', lw=3, ls='--', label='NASDAQ')
    plt.plot(rolling_vol.index, rolling_vol['DIA'], color='red', lw=3, ls=':', label='DOW')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig('fig4_ann_vol.png', dpi=300, bbox_inches='tight')
    plt.show()
    # Rolling volatility summary
    target_cols = stocks + ['SPY', 'QQQ', 'DIA']
    vol_summary = rolling_vol[target_cols].dropna()
    print("\n" + "="*50)
    print("ROLLING VOLATILITY (252-DAY) STATISTICAL SUMMARY")
    print("="*50)
    summary_df = pd.DataFrame({
        'Mean_Vol': vol_summary.mean(),
        'Max_Vol': vol_summary.max(),
        'Min_Vol': vol_summary.min(),
        'Current_Vol': vol_summary.iloc[-1],
        'Vol_Stability': vol_summary.std()
    })
    print(summary_df.round(3).to_markdown())
    print("="*50)

# =====================================================================
# ADIM 1: EVT, TAIL INDEX VE L-MOMENTS ANALİZİ (Ayrı Modül)
# =====================================================================

def calculate_l_moments(x):
    """Probability Weighted Moments (PWM) tabanlı dirençli L-Moments"""
    n = len(x)
    if n < 4:
        return np.nan, np.nan
    
    x_sorted = np.sort(x)
    i = np.arange(1, n + 1)
    
    b0 = np.mean(x_sorted)
    b1 = np.mean(x_sorted * ((i - 1) / (n - 1)))
    b2 = np.mean(x_sorted * ((i - 1) * (i - 2) / ((n - 1) * (n - 2))))
    b3 = np.mean(x_sorted * ((i - 1) * (i - 2) * (i - 3) / ((n - 1) * (n - 2) * (n - 3))))

    l1 = b0
    l2 = 2 * b1 - b0
    l3 = 6 * b2 - 6 * b1 + b0
    l4 = 20 * b3 - 30 * b2 + 12 * b1 - b0

    l_skewness = l3 / l2 if l2 > 0 else np.nan 
    l_kurtosis = l4 / l2 if l2 > 0 else np.nan 

    return l_skewness, l_kurtosis

def calculate_hill_estimator(data, threshold_percentile=95):
    """EVT tabanlı Hill Estimator (Sağ ve Sol Kuyruk Kalınlığı)"""
    data = np.array(data)
    
    # Sağ Kuyruk (Kâr)
    threshold_right = np.percentile(data, threshold_percentile)
    right_tail = data[data > threshold_right]
    if len(right_tail) > 1 and threshold_right > 0:
        gamma_right = np.mean(np.log(right_tail / threshold_right))
        tail_index_right = 1 / gamma_right if gamma_right > 0 else np.nan
    else:
        tail_index_right = np.nan

    # Sol Kuyruk (Zarar)
    data_inv = -data
    threshold_left = np.percentile(data_inv, threshold_percentile)
    left_tail = data_inv[data_inv > threshold_left]
    if len(left_tail) > 1 and threshold_left > 0:
        gamma_left = np.mean(np.log(left_tail / threshold_left))
        tail_index_left = 1 / gamma_left if gamma_left > 0 else np.nan
    else:
        tail_index_left = np.nan

    return tail_index_right, tail_index_left

def extreme_value_tail_analysis(returns_dataframe):
    """
    Ana fonksiyon: Verilen getiri dataframe'indeki tüm kolonlara 
    GARCH uygular ve EVT/Tail analizini çıkarır.
    """
    filtered = {}
    print("Applying AR(1)-GARCH(1,1) filtering for Tail Analysis...")
    for ticker in returns_dataframe.columns:
        filtered[ticker] = apply_garch_filter(returns_dataframe[ticker])
        
    results = []
    print("Executing Extreme Value Theory (EVT) and L-Moments Analysis...")
    
    for ticker, z_scores in filtered.items():
        if len(z_scores) < 100:
            continue
            
        std_skew = stats.skew(z_scores)
        std_kurt = stats.kurtosis(z_scores)
        l_skew, l_kurt = calculate_l_moments(z_scores)
        tail_idx_r, tail_idx_l = calculate_hill_estimator(z_scores, 95)
        tail_asymmetry = tail_idx_l / tail_idx_r if (tail_idx_l and tail_idx_r) else np.nan
        
        max_loss = np.min(z_scores)
        max_gain = np.max(z_scores)
        
        results.append({
            'Asset': ticker,
            'Z_Min(Loss)': max_loss,
            'Z_Max(Gain)': max_gain,
            'Std_Skew': std_skew,
            'L_Skew': l_skew,
            'Std_Kurt': std_kurt,
            'L_Kurt': l_kurt,
            'Tail_Idx_Right': tail_idx_r,
            'Tail_Idx_Left': tail_idx_l,
            'Tail_Asym': tail_asymmetry
        })
        
    df_evt = pd.DataFrame(results).round(4)
    
    print("\n" + "="*95)
    print("EXTREME VALUE THEORY (EVT) & TAIL DYNAMICS (IID Z-Scores)")
    print("="*95)
    print(df_evt.to_markdown(index=False))
    
    print("\n" + "-"*30 + " LaTeX CODE " + "-"*30)
    latex_evt = df_evt.to_latex(index=False, 
                                caption="Extreme Value Theory (EVT) and L-Moments Analysis on IID Z-Scores", 
                                label="tab:evt_analysis")
    print(latex_evt)
    print("-" * 72)
    
    return df_evt

# -------------------------------------------------------------
# ÇALIŞTIRMA
# -------------------------------------------------------------
df_results, returns_df = financial_analysis_pipeline("market_data.csv")

# Tabloyu formatla
report_df = df_results.copy()
for col in report_df.columns:
    if col.endswith('_p'):
        report_df[col] = report_df[col].apply(lambda x: format_sci_values(x, True))
    elif col in ['Corr_R', 'Ann_Vol']:
        report_df[col] = report_df[col].apply(lambda x: format_sci_values(x, False))

print("\n### FINAL ACADEMIC TABLE")
print(report_df.to_markdown(index=False))

print("\n### LATEX CODE")
print(report_df.to_latex(index=False, caption="Statistical Test Results", label="tab:results"))

plot_comprehensive_visuals(df_results, returns_df)

# Kümülatif getiriler
print("\nCumulative Log Returns (End of Period):")
for ticker in all_tickers:
    cum_return = returns_df[ticker].cumsum().iloc[-1]
    print(f"{ticker}: {cum_return:.3f}")

# Performans sıralaması
cum_returns_dict = {ticker: returns_df[ticker].cumsum().iloc[-1] for ticker in all_tickers}
sorted_performance = sorted(cum_returns_dict.items(), key=lambda x: x[1], reverse=True)
print("\nPerformance Ranking:")
for rank, (ticker, ret) in enumerate(sorted_performance, 1):
    print(f"{rank}. {ticker}: {ret:.3f}")

# Beta hesaplamaları
print("\nBeta Calculations vs SPY:")
for stock in stocks:
    if stock in returns_df.columns and 'SPY' in returns_df.columns:
        covariance = np.cov(returns_df[stock], returns_df['SPY'])[0,1]
        variance = np.var(returns_df['SPY'])
        beta = covariance / variance if variance > 0 else np.nan
        print(f"{stock} beta: {beta:.2f}")


# EVT, TAIL INDEX VE L-MOMENTS ANALİZİ Çalıştırma
evt_results_df = extreme_value_tail_analysis(returns_df)        
```

```python
import yfinance as yf
import pandas as pd

def download_market_sector_data():
    # 1. Assets and their corresponding Sector ETFs
    # AAPL/TSLA -> Technology (XLK) / Consumer Disc. (XLY)
    # JPM -> Financials (XLF)
    # XOM -> Energy (XLE)
    # JNJ -> Healthcare (XLV)
    # WMT -> Consumer Staples (XLP)
    
    asset_map = {
        'AAPL': 'XLK',  # Technology
        'TSLA': 'XLY',  # Consumer Discretionary
        'JPM':  'XLF',  # Financials
        'XOM':  'XLE',  # Energy
        'JNJ':  'XLV',  # Healthcare
        'WMT':  'XLP'   # Consumer Staples
    }
    
    # 2. Market Indices (Benchmarks)
    benchmarks = ['SPY', 'QQQ', 'DIA']
    
    # All unique tickers to download
    all_tickers = list(asset_map.keys()) + list(set(asset_map.values())) + benchmarks
    
    print(f"Downloading data for: {all_tickers}")
    
    # 3. Download 5 years of daily data
    # Period: Last 5 years to capture various market regimes
    data = yf.download(all_tickers, period="5y", interval="1d")['Close']
    
    # Handle potential missing values
    data = data.ffill().dropna()
    
    # 4. Save to CSV
    file_name = "market_sector_data.csv"
    data.to_csv(file_name)
    print(f"Successfully saved {len(data)} rows to {file_name}")
    
    # Save the mapping for the analysis step
    mapping_df = pd.DataFrame(list(asset_map.items()), columns=['Stock', 'Sector_ETF'])
    mapping_df.to_csv("sector_mapping.csv", index=False)

if __name__ == "__main__":
    download_market_sector_data()
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.stats.diagnostic import acorr_ljungbox
from sklearn.cluster import AgglomerativeClustering
from arch import arch_model
from sklearn.cluster import KMeans
from scipy.stats import gaussian_kde
import warnings

# Gereksiz uyarıları kapat
warnings.filterwarnings('ignore')

def apply_garch_filter(series):
    """AR(1)-GARCH(1,1) modeli ile seriyi IID hale getirir."""
    scaled = series * 100
    try:
        model = arch_model(scaled, mean='AR', lags=1, vol='GARCH', p=1, q=1, dist='StudentST')
        res = model.fit(disp='off', show_warning=False)
        std_resid = res.std_resid
        return std_resid.dropna().values
    except Exception:
        s = series.values
        return (s - np.mean(s)) / np.std(s)

def format_academic(val, is_p_value=True):
    """Akademik standartlarda yıldızlı p-değeri ve sayı formatlama."""
    if pd.isna(val): return "N/A"
    if is_p_value:
        star = ""
        if val < 0.01: star = "***"
        elif val < 0.05: star = "**"
        elif val < 0.10: star = "*"
        if val < 0.001: return f"<.001{star}"
        formatted = f"{val:.3f}".lstrip('0')
        if not formatted.startswith('.'): formatted = '0' + formatted
        return f"{formatted}{star}"
    else:
        return f"{val:.4f}"

def stage_2_sector_analysis_complete(data_path, mapping_path):
    # 1. VERİ YÜKLEME
    df = pd.read_csv(data_path, index_col=0, parse_dates=True)
    mapping = pd.read_csv(mapping_path)
    returns = np.log(df / df.shift(1)).dropna()
    
    # Tüm benzersiz sembolleri topla ve filtreden geçir
    all_symbols = list(set(mapping['Stock']).union(set(mapping['Sector_ETF'])))
    filtered = {}
    print("Applying AR(1)-GARCH(1,1) filtering to sectoral data...")
    for sym in all_symbols:
        if sym in returns.columns:
            filtered[sym] = apply_garch_filter(returns[sym])
    
    results_list = []
    
    # 2. ANALİZ DÖNGÜSÜ
    for _, row in mapping.iterrows():
        stock = row['Stock']
        sector_etf = row['Sector_ETF']
        
        if stock not in filtered or sector_etf not in filtered:
            continue
            
        s_std_garch = filtered[stock]
        r_std_garch = filtered[sector_etf]
        
        # Hizalama
        min_len = min(len(s_std_garch), len(r_std_garch))
        s_std_garch = s_std_garch[-min_len:]
        r_std_garch = r_std_garch[-min_len:]
        
        s_raw_aligned = returns[stock].values[-min_len:]
        r_raw_aligned = returns[sector_etf].values[-min_len:]
        
        s_std = (s_std_garch - np.mean(s_std_garch)) / np.std(s_std_garch)
        r_std = (r_std_garch - np.mean(r_std_garch)) / np.std(r_std_garch)
        
        # Testler
        ed_total = stats.energy_distance(s_raw_aligned, r_raw_aligned)
        ed_shape = stats.energy_distance(s_std, r_std)
        ks_p = stats.ks_2samp(s_std, r_std).pvalue
        cvm_p = stats.cramervonmises_2samp(s_std, r_std).pvalue
        try:
            es_p = stats.epps_singleton_2samp(s_std, r_std)[1]
        except:
            es_p = np.nan
            
        jb_p = stats.jarque_bera(s_raw_aligned)[1]
        corr = np.corrcoef(s_raw_aligned, r_raw_aligned)[0, 1]
        
        if len(s_std) > 10:
            lb_p = acorr_ljungbox(s_std, lags=[10], return_df=True).iloc[0,1]
        else:
            lb_p = np.nan
        
        results_list.append({
            'Stock': stock, 'Sector': sector_etf, 'Corr': corr, 'ED_Total': ed_total, 'ED_Shape': ed_shape,
            'LB_p': lb_p, 'KS_p': ks_p,
            'CvM_p': cvm_p, 'ES_p': es_p, 
            'JB_p': jb_p,
        })
    
    df_results = pd.DataFrame(results_list)

    # 3. TABLO FORMATLAMA
    report_df = df_results.copy()
    p_cols = ['JB_p', 'LB_p', 'KS_p', 'CvM_p', 'ES_p']
    m_cols = ['Corr', 'ED_Total', 'ED_Shape'] 

    for col in p_cols:
        report_df[col] = report_df[col].apply(lambda x: format_academic(x, True))
    for col in m_cols:
        report_df[col] = report_df[col].apply(lambda x: format_academic(x, False))

    print("\n" + "="*100)
    print("STAGE 2: FINAL SECTORAL SIMILARITY RESULTS (Dual Energy Distance & IID Adjusted)")
    print("="*100)
    print(report_df.to_markdown(index=False, disable_numparse=True))

    print("\n" + "-"*30 + " LaTeX CODE " + "-"*30)
    latex_table = report_df.to_latex(index=False, 
                                     caption="Distributional Similarity Analysis: Total vs. Structural Distance (ARMA-GARCH Filtered)", 
                                     label="tab:sector_analysis_dual",
                                     column_format="llcccccccc",
                                     escape=False)
    print(latex_table)
    print("-" * 72)

    # 4. GÖRSELLEŞTİRME
    sns.set_theme(style="white")
    
    plt.figure(figsize=(10, 5))
    heatmap_data = df_results.set_index('Stock')[['KS_p', 'CvM_p', 'ES_p']]
    sns.heatmap(heatmap_data, annot=True, cmap="RdYlGn", center=0.05, fmt=".3f",
                cbar_kws={'label': 'P-Value (Similarity)'})
    plt.tight_layout()
    plt.savefig('fig5_stg2_heatmap.png', dpi=300, bbox_inches='tight')
    plt.show()

    # --- DÜZELTİLMİŞ KDE VERİ YAZDIRMA ---
    print("\n" + "="*80)
    print("KDE INTERPRETATION DATA (Standardized GARCH Residuals)")
    print("="*80)
    
    for _, row in mapping.iterrows():
        stock = row['Stock']
        sector_etf = row['Sector_ETF']
        
        if stock not in filtered or sector_etf not in filtered:
            continue
            
        s_garch = filtered[stock]
        r_garch = filtered[sector_etf]
        min_len = min(len(s_garch), len(r_garch)) # Kapsam hatası giderildi
        
        s_std = s_garch[-min_len:]
        r_std = r_garch[-min_len:]
        
        s_std = (s_std - np.mean(s_std)) / np.std(s_std)
        r_std = (r_std - np.mean(r_std)) / np.std(r_std)
        
        # Güvenli Pandas mode kullanımı
        #s_peak = pd.Series(np.round(s_std, 1)).mode()[0] if len(s_std) > 0 else np.nan
        #r_peak = pd.Series(np.round(r_std, 1)).mode()[0] if len(r_std) > 0 else np.nan
        
        # Mode yerine Gaussian KDE ile gerçek zirveyi (peak) bulma
        kde_s = gaussian_kde(s_std)
        kde_r = gaussian_kde(r_std)
        
        # -10 ile 10 arasında 1000 noktalık bir ızgara oluştur
        x_grid = np.linspace(-10, 10, 1000)
        
        s_peak = x_grid[np.argmax(kde_s(x_grid))]
        r_peak = x_grid[np.argmax(kde_r(x_grid))]
        
        print(f"\n{stock} vs {sector_etf}:")
        print(f"  Stock - Peak at {s_peak:.3f}, Range: [{s_std.min():.2f}, {s_std.max():.2f}]")
        print(f"  ETF   - Peak at {r_peak:.3f}, Range: [{r_std.min():.2f}, {r_std.max():.2f}]")
        print(f"  Peak difference: {abs(s_peak - r_peak):.3f}")
        print(f"  KS p-value: {stats.ks_2samp(s_std, r_std).pvalue:.3f}")
        print(f"  ED_Shape: {stats.energy_distance(s_std, r_std):.4f}")
    
    # --- DÜZELTİLMİŞ KDE GRAFİKLERİ ---
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    axes = axes.flatten()
    for i, (_, row) in enumerate(mapping.iterrows()):
        stock = row['Stock']
        sector_etf = row['Sector_ETF']
        
        if stock not in filtered or sector_etf not in filtered:
            continue
            
        s_garch = filtered[stock]
        r_garch = filtered[sector_etf]
        min_len = min(len(s_garch), len(r_garch)) # Kapsam hatası giderildi
        
        s_std = s_garch[-min_len:]
        r_std = r_garch[-min_len:]
        
        s_std = (s_std - np.mean(s_std)) / np.std(s_std)
        r_std = (r_std - np.mean(r_std)) / np.std(r_std)
        
        sns.kdeplot(s_std, ax=axes[i], label=f"{stock} (IID)", fill=True, alpha=0.2)
        sns.kdeplot(r_std, ax=axes[i], label=f"{sector_etf} (IID)", ls='--', color='darkred', lw=2)
        axes[i].set_title(f"{stock} vs {sector_etf}")
        axes[i].legend()

    plt.tight_layout()
    plt.savefig('fig6_overlap.png', dpi=300, bbox_inches='tight')
    plt.show()

    return returns, report_df

# =====================================================================
# ADIM 1: EVT, TAIL INDEX VE L-MOMENTS ANALİZİ (Ayrı Modül)
# =====================================================================

def calculate_l_moments(x):
    """Probability Weighted Moments (PWM) tabanlı dirençli L-Moments"""
    n = len(x)
    if n < 4:
        return np.nan, np.nan
    
    x_sorted = np.sort(x)
    i = np.arange(1, n + 1)
    
    b0 = np.mean(x_sorted)
    b1 = np.mean(x_sorted * ((i - 1) / (n - 1)))
    b2 = np.mean(x_sorted * ((i - 1) * (i - 2) / ((n - 1) * (n - 2))))
    b3 = np.mean(x_sorted * ((i - 1) * (i - 2) * (i - 3) / ((n - 1) * (n - 2) * (n - 3))))

    l1 = b0
    l2 = 2 * b1 - b0
    l3 = 6 * b2 - 6 * b1 + b0
    l4 = 20 * b3 - 30 * b2 + 12 * b1 - b0

    l_skewness = l3 / l2 if l2 > 0 else np.nan 
    l_kurtosis = l4 / l2 if l2 > 0 else np.nan 

    return l_skewness, l_kurtosis

def calculate_hill_estimator(data, threshold_percentile=95):
    """EVT tabanlı Hill Estimator (Sağ ve Sol Kuyruk Kalınlığı)"""
    data = np.array(data)
    
    # Sağ Kuyruk (Kâr)
    threshold_right = np.percentile(data, threshold_percentile)
    right_tail = data[data > threshold_right]
    if len(right_tail) > 1 and threshold_right > 0:
        gamma_right = np.mean(np.log(right_tail / threshold_right))
        tail_index_right = 1 / gamma_right if gamma_right > 0 else np.nan
    else:
        tail_index_right = np.nan

    # Sol Kuyruk (Zarar)
    data_inv = -data
    threshold_left = np.percentile(data_inv, threshold_percentile)
    left_tail = data_inv[data_inv > threshold_left]
    if len(left_tail) > 1 and threshold_left > 0:
        gamma_left = np.mean(np.log(left_tail / threshold_left))
        tail_index_left = 1 / gamma_left if gamma_left > 0 else np.nan
    else:
        tail_index_left = np.nan

    return tail_index_right, tail_index_left

def extreme_value_tail_analysis(returns_dataframe):
    """
    Ana fonksiyon: Verilen getiri dataframe'indeki tüm kolonlara 
    GARCH uygular ve EVT/Tail analizini çıkarır.
    """
    filtered = {}
    print("Applying AR(1)-GARCH(1,1) filtering for Tail Analysis...")
    for ticker in returns_dataframe.columns:
        filtered[ticker] = apply_garch_filter(returns_dataframe[ticker])
        
    results = []
    print("Executing Extreme Value Theory (EVT) and L-Moments Analysis...")
    
    # Fonksiyonun içindeki döngü kısmı:
    for ticker, z_scores in filtered.items():
        if len(z_scores) < 100:
            continue
            
        std_skew = stats.skew(z_scores)
        std_kurt = stats.kurtosis(z_scores)
        l_skew, l_kurt = calculate_l_moments(z_scores)
        
        # HEM %5 HEM %1 İÇİN HESAPLAMA:
        tail_idx_r_5, tail_idx_l_5 = calculate_hill_estimator(z_scores, 95)
        tail_idx_r_1, tail_idx_l_1 = calculate_hill_estimator(z_scores, 99)
        
        tail_asymmetry = tail_idx_l_5 / tail_idx_r_5 if (tail_idx_l_5 and tail_idx_r_5) else np.nan
        
        max_loss = np.min(z_scores)
        max_gain = np.max(z_scores)
        
        results.append({
            'Asset': ticker,
            'Z_Min(Loss)': max_loss,
            'Std_Skew': std_skew,
            'L_Skew': l_skew,
            'L_Kurt': l_kurt,
            'Tail_L_5%': tail_idx_l_5,  # 5% Sol Kuyruk
            'Tail_R_5%': tail_idx_r_5,  # 5% Sağ Kuyruk
            'Tail_L_1%': tail_idx_l_1,  # 1% Sol Kuyruk (HAKEM İSTEĞİ)
            'Tail_R_1%': tail_idx_r_1,  # 1% Sağ Kuyruk
            'Tail_Asym': tail_asymmetry
        })
        
    df_evt = pd.DataFrame(results).round(4)
    
    print("\n" + "="*95)
    print("EXTREME VALUE THEORY (EVT) & TAIL DYNAMICS (IID Z-Scores)")
    print("="*95)
    print(df_evt.to_markdown(index=False))
    
    print("\n" + "-"*30 + " LaTeX CODE " + "-"*30)
    latex_evt = df_evt.to_latex(index=False, 
                                caption="Extreme Value Theory (EVT) and L-Moments Analysis on IID Z-Scores", 
                                label="tab:evt_analysis")
    print(latex_evt)
    print("-" * 72)
    
    return df_evt
    
# =====================================================================
# ADIM 3: ECONOMIC SIGNIFICANCE & 3-TIER BOOTSTRAPPED PORTFOLIOS
# =====================================================================

def stage_3_bootstrapped_backtest(returns_df, stage2_results_df, evt_results_df, num_portfolios=1000):
    print("\n" + "="*80)
    print("STAGE 3: RUNNING 3-TIER BOOTSTRAPPED PORTFOLIO SIMULATION")
    print("="*80)
    
    # 1. Veri Hazırlığı
    stocks = stage2_results_df['Stock'].tolist()
    stock_returns = returns_df[stocks].copy()
    
    # Metrikleri Çekme
    ed_values = stage2_results_df['ED_Shape'].astype(float).values
    ks_p_raw = stage2_results_df['KS_p'].str.replace(r'[^\d.]', '', regex=True).astype(float).values
    
    # EVT verisini hisse sırasına göre eşleştir (Z_Min Loss değerini al)
    evt_matched = evt_results_df.set_index('Asset').loc[stocks].reset_index()
    z_min_loss = evt_matched['Z_Min(Loss)'].astype(float).values
    
    # 2. ÜÇ FARKLI MODEL İÇİN AĞIRLIKLANDIRMA
    
     # EVT verisini hisse sırasına göre eşleştir (L-Kurtosis değerini al)
    evt_matched = evt_results_df.set_index('Asset').loc[stocks].reset_index()
    l_kurt_vals = evt_matched['L_Kurt'].astype(float).values
    
    # MODEL 1: Sadece Energy Distance (Inverse ED)
    m1_raw = 1.0 / ed_values
    w_model1 = m1_raw / np.sum(m1_raw)
    
    # MODEL 2: Multi-Factor (ED + KS_p)
    m2_raw = (1.0 / ed_values) * (ks_p_raw + 0.001)
    w_model2 = m2_raw / np.sum(m2_raw)
    
    # MODEL 3: Ultimate EVT-KDE Synthesis (ED + KS_p + L-Kurtosis Penalty)
    # L_Kurt yapısal kuyruk riskidir. Yüksek olması kötüdür, bu yüzden paydaya alarak cezalandırıyoruz.
    m3_raw = ((1.0 / ed_values) * (ks_p_raw + 0.001)) / l_kurt_vals
    w_model3 = m3_raw / np.sum(m3_raw)
    
    # Portföy Getirileri ve Sharpe Oranları (Yıllıklandırılmış)
    def calc_sharpe(weights):
        port_ret = stock_returns.dot(weights)
        return (port_ret.mean() / port_ret.std()) * np.sqrt(252)

    # YENİ: Expected Shortfall (CVaR) - %5 en kötü durumların ortalaması
    def calc_expected_shortfall(weights, confidence_level=0.05):
        port_ret = stock_returns.dot(weights)
        threshold = np.percentile(port_ret, confidence_level * 100)
        cvar = port_ret[port_ret <= threshold].mean()
        return cvar * 100 # Yüzde cinsinden göstermek için

    sharpe_m1 = calc_sharpe(w_model1)
    sharpe_m2 = calc_sharpe(w_model2)
    sharpe_m3 = calc_sharpe(w_model3)
    
    es_m1 = calc_expected_shortfall(w_model1)
    es_m2 = calc_expected_shortfall(w_model2)
    es_m3 = calc_expected_shortfall(w_model3)
    
    # 3. Bootstrapping: 1000 Rastgele Portföy
    np.random.seed(42)
    random_sharpes = []
    
    for _ in range(num_portfolios):
        rand_w = np.random.dirichlet(np.ones(len(stocks)))
        rand_sharpe = calc_sharpe(rand_w)
        random_sharpes.append(rand_sharpe)
        
    random_sharpes = np.array(random_sharpes)
    
    # Performans Yüzdeleri
    beat_m1 = np.sum(sharpe_m1 > random_sharpes) / num_portfolios * 100
    beat_m2 = np.sum(sharpe_m2 > random_sharpes) / num_portfolios * 100
    beat_m3 = np.sum(sharpe_m3 > random_sharpes) / num_portfolios * 100
    
    # 4. Raporlama
    # Raporlama kısmını güncelle:
    print(f"Random Portfolios Mean Sharpe : {random_sharpes.mean():.3f}")
    print("-" * 65)
    print(f"MODEL 1 (ED Only)           -> Sharpe: {sharpe_m1:.3f} (Beats {beat_m1:.1f}%)| Expected Shortfall: {es_m1:.2f}%")
    print(f"MODEL 2 (ED + KS)           -> Sharpe: {sharpe_m2:.3f} (Beats {beat_m2:.1f}%)| Expected Shortfall: {es_m2:.2f}%")
    print(f"MODEL 3 (ED+KS+L-Kurtosis)  -> Sharpe: {sharpe_m3:.3f} (Beats {beat_m3:.1f}%)| Expected Shortfall: {es_m3:.2f}%")
    print("*(Lower Expected Shortfall indicates better tail-risk protection)*")
    
    # 5. Görselleştirme (Hepsi Tek Grafikte)
    sns.set_theme(style="whitegrid")
    plt.figure(figsize=(12, 7))
    
    # Rastgele portföylerin Çan Eğrisi
    sns.kdeplot(random_sharpes, fill=True, color='lightgray', alpha=0.6, label='Bootstrapped Random Portfolios')
    
    # Ortalama Çizgisi
    plt.axvline(x=random_sharpes.mean(), color='black', linestyle=':', linewidth=1.5, label='Random Mean')
    
    # Model Çizgileri
    plt.axvline(x=sharpe_m1, color='blue', linestyle='--', linewidth=2, label=f'Model 1: ED-Only (Sharpe: {sharpe_m1:.3f})')
    plt.axvline(x=sharpe_m2, color='orange', linestyle='--', linewidth=2, label=f'Model 2: ED+KS (Sharpe: {sharpe_m2:.3f})')
    plt.axvline(x=sharpe_m3, color='red', linestyle='-', linewidth=2.5, label=f'Model 3: EVT-KDE Synthesis (Sharpe: {sharpe_m3:.3f})')
    
    plt.title('Evolution of Portfolio Performance Using Structural & EVT Metrics', fontsize=14, fontweight='bold')
    plt.xlabel('Annualized Sharpe Ratio', fontsize=12)
    plt.ylabel('Density', fontsize=12)
    plt.legend(loc='upper right')
    
    plt.tight_layout()
    plt.savefig('fig7_3_tier_backtest.png', dpi=300, bbox_inches='tight')
    plt.show()

    return w_model1, w_model2, w_model3, random_sharpes
# =====================================================================
# ADIM 4: HIERARCHICAL CLUSTERING (Hakemin "Clustering" Talebine Yanıt)
# =====================================================================


def stage_4_structural_clustering(stage2_df, evt_df):
    print("\n" + "="*80)
    print("STAGE 4: HIERARCHICAL CLUSTERING (Structural Divergence vs. Tail Risk)")
    print("="*80)

    # 1. Verileri Birleştirme
    df_ed = stage2_df[['Stock', 'ED_Shape']].copy()
    df_ed['ED_Shape'] = df_ed['ED_Shape'].astype(float)
    
    df_evt = evt_df[['Asset', 'L_Kurt']].copy()
    df_evt['L_Kurt'] = df_evt['L_Kurt'].astype(float)
    df_evt.rename(columns={'Asset': 'Stock'}, inplace=True)
    
    cluster_df = pd.merge(df_ed, df_evt, on='Stock')
    
    # 2. Hiyerarşik Kümeleme (Agglomerative)
    X = cluster_df[['ED_Shape', 'L_Kurt']].values
    
    # Hakemin istediği Hiyerarşik yaklaşım
    hc = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
    cluster_df['Cluster'] = hc.fit_predict(X)
    
    # Hiyerarşik kümelemede cluster_centers_ olmadığı için manuel hesaplıyoruz:
    centers = cluster_df.groupby('Cluster')[['ED_Shape', 'L_Kurt']].mean()
    risk_scores = centers.sum(axis=1) # Toplam risk skoru
    
    # Riskleri sırala
    sorted_clusters = risk_scores.sort_values().index
    
    cluster_mapping = {
        sorted_clusters[0]: 'Cluster 1 (Safe Haven / Low Risk)',
        sorted_clusters[1]: 'Cluster 2 (Gray Zone / Moderate)',
        sorted_clusters[2]: 'Cluster 3 (Toxic / High Risk)'
    }
    
    cluster_df['Profile'] = cluster_df['Cluster'].map(cluster_mapping)
    
    # 3. Görselleştirme (2D Scatter Plot)
    plt.figure(figsize=(10, 6))
    sns.set_theme(style="whitegrid")
    
    # Renk Paleti: Güvenli=Yeşil, Orta=Turuncu, Riskli=Kırmızı
    palette = {'Cluster 1 (Safe Haven / Low Risk)': '#2ca02c', 
               'Cluster 2 (Gray Zone / Moderate)': '#ff7f0e', 
               'Cluster 3 (Toxic / High Risk)': '#d62728'}
               
    ax = sns.scatterplot(data=cluster_df, x='ED_Shape', y='L_Kurt', hue='Profile', 
                         palette=palette, s=150, edgecolor='black', alpha=0.8)
    
    # Hisselerin isimlerini noktaların yanına yazdır (etiketleme)
    for i in range(len(cluster_df)):
        plt.text(cluster_df['ED_Shape'][i] + 0.001, cluster_df['L_Kurt'][i] + 0.001, 
                 cluster_df['Stock'][i], fontsize=9, weight='bold', color='black')
                 
    plt.title('Hierarchical Clustering: Structural Divergence vs. Tail Risk', fontsize=14, fontweight='bold')
    plt.xlabel('Structural Divergence (ED_Shape)', fontsize=12)
    plt.ylabel('Tail Risk (L-Kurtosis)', fontsize=12)
    plt.legend(title='Risk Profiles', loc='upper left')
    
    plt.tight_layout()
    plt.savefig('fig8_clustering.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print("Clustering completed! Check the 'fig8_clustering.png' file.")
    return cluster_df
    
# =====================================================================
# ÇALIŞTIRMA BLOĞU (Tüm Aşamalar)
# =====================================================================

# Stage 2: Sektörel Analiz, p-değerleri ve Energy Distance sonuçlarını al
returns_stage2_df, stage2_results_df = stage_2_sector_analysis_complete("market_sector_data.csv", "sector_mapping.csv")

# EVT (Kuyruk Analizi): Hakemin "Kuyruk Asimetrisi" talebine yanıt
evt_results_stage2 = extreme_value_tail_analysis(returns_stage2_df)

# Stage 3 (YENİ): Hakemin "Bootstrapping ve Yapısal Önyargı" talebine yanıt
# Stage 2'den gelen tabloyu (stage2_results_df) kullanarak portföy simülasyonunu başlat
w1, w2, w3, rand_sims = stage_3_bootstrapped_backtest(returns_stage2_df, stage2_results_df, evt_results_stage2, num_portfolios=1000)

# STAGE 4 ÇALIŞTIRMA (Kümeleme Algoritması)
cluster_results_df = stage_4_structural_clustering(stage2_results_df, evt_results_stage2)
```
